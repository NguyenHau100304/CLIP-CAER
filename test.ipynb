```python
import random
import os

def undersample_dataset(input_file, output_file, class_1_target=500):
    """
    Hàm giảm số lượng mẫu Class 1 xuống còn class_1_target mẫu.
    Các lớp khác giữ nguyên.
    """
    if not os.path.exists(input_file):
        print(f"Lỗi: Không tìm thấy file {input_file}")
        return

    with open(input_file, 'r') as f:
        lines = f.readlines()

    class_1_lines = []
    other_lines = []

    # Phân loại các dòng
    for line in lines:
        parts = line.strip().split()
        if len(parts) < 3: continue
        
        # Label nằm ở vị trí cuối cùng (index 2)
        label = int(parts[2]) 
        
        if label == 1:  # Nếu là Class 1
            class_1_lines.append(line)
        else:           # Các Class còn lại (2, 3, 4, 5)
            other_lines.append(line)

    # In thống kê trước khi giảm
    print(f"--- Trước khi Undersample ---")
    print(f"Class 1: {len(class_1_lines)}")
    print(f"Các Class khác: {len(other_lines)}")
    print(f"Tổng cộng: {len(lines)}")

    # Thực hiện Undersample Class 1
    # Trộn ngẫu nhiên để không bị thiên lệch theo thứ tự video
    random.seed(42) 
    random.shuffle(class_1_lines)
    
    # Chỉ lấy số lượng mong muốn (ví dụ 500 mẫu)
    # 500 là con số hợp lý, tương đương với Class 5 (422 mẫu)
    keep_class_1 = class_1_lines[:class_1_target]
    
    # Gộp lại thành danh sách mới
    new_data = keep_class_1 + other_lines
    random.shuffle(new_data) # Trộn đều tất cả lần cuối

    # Lưu ra file mới
    with open(output_file, 'w') as f:
        f.writelines(new_data)

    print(f"\n--- Sau khi Undersample ---")
    print(f"Class 1 (giữ lại): {len(keep_class_1)}")
    print(f"Các Class khác: {len(other_lines)}")
    print(f"Tổng cộng mới: {len(new_data)}")
    print(f"\nĐã lưu file mới tại: {output_file}")

# --- CẤU HÌNH ---
# Đường dẫn file train gốc
INPUT_TXT = 'annotation/train_new.txt'
# Đường dẫn file train mới sẽ tạo ra
OUTPUT_TXT = 'annotation/train_undersampled.txt'

# Số lượng mẫu Class 1 muốn giữ lại (nên để tầm 450-500 cho cân với Class 5)
TARGET_NUM = 500 

undersample_dataset(INPUT_TXT, OUTPUT_TXT, TARGET_NUM)
```
